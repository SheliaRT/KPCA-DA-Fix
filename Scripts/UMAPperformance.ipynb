### ComBat on UMAP

```python
#import dataset
#SOURCE
expr_file1= r"C:\Zmyfiles\codes\research\KPCA-DA\KPCA-DA\examples\Data\TvsN\TCGA_BRCA_NvT.csv"
expr1 = pd.read_csv(expr_file1, encoding='utf-8', engine='python')

#TARGET_test_unadjusted
expr_file2= r"C:\Zmyfiles\codes\research\KPCA-DA\KPCA-DA\examples\Data\TvsN\target_test.csv"
expr2 = pd.read_csv(expr_file2, encoding='utf-8', engine='python')

#TARGET_test_adjusted
expr_file3= r"C:\Zmyfiles\codes\research\KPCA-DA\KPCA-DA\examples\Data\TvsN\aligned_tcga.csv"
expr_file3 = pd.read_csv(expr_file3, encoding='utf-8', engine='python')

expr_file4= r"C:\Zmyfiles\codes\research\KPCA-DA\KPCA-DA\examples\Data\TvsN\aligned_gse.csv"
expr4 = pd.read_csv(expr_file4, encoding='utf-8', engine='python')
```

```python
# Check the shapes of the datasets
print(f"expr1 shape: {expr1.shape}")
print(f"expr2 shape: {expr2.shape}")
print(f"expr_file3 shape: {expr_file3.shape}")
print(f"\nAfter transpose:")
print(f"expr1_umap would be: {expr1.drop('Hugo_Symbol', axis=1).T.shape}")
print(f"expr2_umap would be: {expr_file3.drop('Hugo_Symbol', axis=1).T.shape}")
```

```python
# Prepare data for UMAP (remove 'Hugo_Symbol' column and transpose so samples are rows)
expr1_umap = expr1.drop('Hugo_Symbol', axis=1).T
expr2_umap = expr2.drop('Hugo_Symbol', axis=1).T

# Fit UMAP on expr1 and transform both datasets
umap_model = umap_.UMAP(n_components=2, random_state=42)
expr1_embedded = umap_model.fit_transform(expr1_umap)
expr2_embedded = umap_model.transform(expr2_umap)

# Plot in separate subplots
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.scatter(expr1_embedded[:, 0], expr1_embedded[:, 1], s=30, alpha=0.6, c='blue')
plt.title('UMAP of expr1 (Source)')
plt.xlabel('UMAP1')
plt.ylabel('UMAP2')

plt.subplot(1, 2, 2)
plt.scatter(expr2_embedded[:, 0], expr2_embedded[:, 1], s=30, alpha=0.6, c='orange')
plt.title('UMAP of expr2 (Target)')
plt.xlabel('UMAP1')
plt.ylabel('UMAP2')

plt.tight_layout()
plt.show()
```

```python
# Prepare data for UMAP (remove 'Hugo_Symbol' column and transpose so samples are rows)
expr1_umap = (expr1.drop('gene', axis=1).iloc[:-1]).T
expr2_umap = expr2.drop('gene', axis=1).T

# Fit UMAP on expr1 and transform both datasets
umap_model = umap_.UMAP(n_components=2)
expr1_embedded = umap_model.fit_transform(expr1_umap)
expr2_embedded = umap_model.transform(expr2_umap)

# Plot both datasets in the same plot
plt.figure(figsize=(10, 8))
plt.scatter(expr1_embedded[:, 0], expr1_embedded[:, 1], s=30, alpha=0.6, label='Source', c='blue')
plt.scatter(expr2_embedded[:, 0], expr2_embedded[:, 1], s=30, alpha=0.6, label='Target', c='orange')
plt.title('UMAP: Adaptation with CAN', fontsize=20)
plt.xlabel('Source Dataset: TCGA', fontsize=20)
plt.ylabel('Target Dataset: GEO', fontsize=20)
plt.legend(fontsize=18)
plt.tight_layout()
plt.show()
```

```python
# m1, m2, disparity = procrustes(expr1_embedded, expr2_embedded)
# print("Procrustes similarity:", 1 - disparity)
emd_x = wasserstein_distance(expr1_embedded[:,0], expr2_embedded[:,0])
emd_y = wasserstein_distance(expr1_embedded[:,1], expr2_embedded[:,1])
#print("Wasserstein distance (x):", emd_x, " (y):", emd_y)

emd_combined = np.sqrt(emd_x**2 + emd_y**2)

print(f"Wasserstein distance X-axis: {emd_x:.4f}")
print(f"Wasserstein distance Y-axis: {emd_y:.4f}")
print(f"Combined Wasserstein distance (scalar): {emd_combined:.4f}")
```

```python
# Prepare data for UMAP (remove 'Hugo_Symbol' column and transpose so samples are rows)
expr3_umap = expr_file3.T
expr4_umap = expr4.drop('Hugo_Symbol', axis=1).T

# Fit UMAP on expr1 and transform both datasets
umap_model2 = umap_.UMAP(n_components=2)
expr3_embedded = umap_model2.fit_transform(expr3_umap)
expr4_embedded = umap_model2.transform(expr4_umap)

# Plot both datasets in the same plot
plt.figure(figsize=(10, 8))
plt.scatter(expr3_embedded[:, 0], expr3_embedded[:, 1], s=30, alpha=0.6, label='Source', c='blue')
plt.scatter(expr4_embedded[:, 0], expr4_embedded[:, 1], s=30, alpha=0.6, label='Target', c='orange')
plt.title('UMAP: Adaptation with ComBat', fontsize=20)
plt.xlabel('Source Dataset: TCGA', fontsize=20)
plt.ylabel('Target Dataset: GEO', fontsize=20)
plt.legend(fontsize=18)
plt.tight_layout()
plt.show()
```
